```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('heart.csv')
```

```python
df.head()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='target', y='age')
plt.title('Boxplot of Age by Heart Disease Status')
plt.xlabel('Heart Disease (1 = Yes, 0 = No)')
plt.ylabel('Age')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(data=df, x='chol', bins=30, kde=True, color='orange')
plt.title('Histogram of Cholesterol Levels')
plt.xlabel('Cholesterol')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
avg_hr = df.groupby('age')['thalach'].mean()
avg_hr.plot(kind='line', marker='o', color='green')
plt.title('Average Maximum Heart Rate by Age')
plt.xlabel('Age')
plt.ylabel('Max Heart Rate (thalach)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
avg_metrics = df.groupby('age')[['chol', 'thalach', 'trestbps']].mean()
avg_metrics.plot(marker='o')
plt.title('Average Cholesterol, Max HR, and Resting BP by Age')
plt.xlabel('Age')
plt.ylabel('Average Values')
plt.grid(True)
plt.tight_layout()
plt.show()
```
