```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('dataset_Facebook.csv', delimiter=';', encoding='latin1')
```

```python
df.head()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='Type', y='like')
plt.title('Distribution of Likes by Post Type')
plt.xlabel('Post Type')
plt.ylabel('Number of Likes')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(data=df, x='Total Interactions', bins=30, kde=True)
plt.title('Histogram of Total Interactions')
plt.xlabel('Total Interactions')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
avg_interactions = df.groupby('Post Hour')['Total Interactions'].mean()
avg_interactions.plot(kind='line', marker='o')
plt.title('Average Total Interactions by Post Hour')
plt.xlabel('Post Hour')
plt.ylabel('Average Total Interactions')
plt.xticks(range(0, 24))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
avg_metrics = df.groupby('Post Hour')[['like', 'comment', 'share']].mean()
avg_metrics.plot(marker='o')
plt.title('Average Likes, Comments, and Shares by Post Hour')
plt.xlabel('Post Hour')
plt.ylabel('Average Count')
plt.xticks(range(0, 24))
plt.grid(True)
plt.tight_layout()
plt.show()
```
