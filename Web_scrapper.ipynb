```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.flipkart.com/apple-iphone-16-pro-desert-titanium-128-gb/p/itm5a8453e89cbd4?pid=MOBH4DQFSDYNVH5U&lid=LSTMOBH4DQFSDYNVH5UFZYHLF&marketplace=FLIPKART&cmpid=content_mobile_22371611432_x_8965229628_gmc_pla&tgi=sem,1,G,11214002,x,,,,,,,c,,,,,,,&&cmpid=content_22371611432_gmc_pla&gad_source=1&gclid=CjwKCAjw5PK_BhBBEiwAL7GTPVLNr1MKfa9m7-Q3jJBoQjbiM1LXgewEL_ZCdp9HKxDEDYQT7PafHxoCfNsQAvD_BwE"
headers ={"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/135.0.0.0 Safari/537.36"}

req = requests.get(url, headers=headers)
soup = BeautifulSoup(req.content,"html.parser")
```

```python
reviews = soup.findAll('div',{'class' : 'ZmyHeo'});
for review in reviews:
    print(review.get_text() + "\n\n")
```

```python
ratings = soup.find('div',{'class':'ipqd2A'}).get_text();
print(ratings)
```

```python
individual_ratings = soup.findAll('div',{'class' : 'XQDdHH Ga3i8K'});
for indi_ratings in individual_ratings:
    print(indi_ratings.get_text() + "\n") 
                                         
```

```python
customer_name = soup.findAll('p' ,{'class' : '_2NsDsF AwS1CA'});
for cust_name in customer_name:
    print(cust_name.get_text() + "\n\n")

```

```python
price = soup.find('div' ,{'class' : 'UOCQB1'}).get_text();
print(price)
```

```python
questions = soup.findAll('span',{'class' : 'wys2hv _43gOsC'});
for question in questions:
    print(question.get_text())
    answer = question.findNext('span',{'class' :'-iXTM3'}).get_text();
    print(answer + "\n\n")
```
